# 07 — LOTO with Rolling Features and Threshold Calibration (Phase 2)

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent) if Path.cwd().name=='notebooks' else str(Path.cwd()))
from src.utils.data_loader import DatasetSource
from src.ml.loto import run_loto
import pandas as pd

src = DatasetSource()
ei = src.read_csv('Wind Farm A/event_info.csv', sep=';')
FEATURES = ['sensor_0_avg','wind_speed_3_avg','sensor_5_avg','power_30_avg','sensor_11_avg','sensor_13_avg']

res = run_loto(src, 'A', FEATURES, ei, use_rolling=True, windows=(6,36), calibrate=True, if_contaminations=[0.05,0.10,0.15], z_threshold=2.0)
res.to_csv('reports/loto_rolling_calibrated.csv', index=False)
print(res.groupby(['model','contamination'])[['precision','recall','f1','false_alarms_per_day','event_detection_rate']].mean().round(4).to_string())
```

```python
# Compare with baseline LOTO (no rolling)
base = pd.read_csv('reports/loto_results.csv')
print('Baseline LOTO z-score F1:', round(base[base['model']=='zscore']['f1'].mean(),4))
print('Rolling+calibrated z-score F1:', round(res[res['model']=='zscore']['f1'].mean(),4))
```
