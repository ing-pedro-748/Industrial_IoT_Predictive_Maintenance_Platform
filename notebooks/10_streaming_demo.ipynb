```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent) if Path.cwd().name=='notebooks' else str(Path.cwd()))
from src.utils.data_loader import DatasetSource
from src.streaming.simulator import replay_series
from src.streaming.consumer import WindowAggregator
import pandas as pd

src = DatasetSource()
df = src.read_csv('Wind Farm A/datasets/0.csv', sep=';')
FEATURES = ['sensor_0_avg','wind_speed_3_avg','sensor_11_avg']

events = []
def publish(ev): events.append(ev)
n = replay_series(df.head(500), 'A', FEATURES, publish, max_events=500)
print('published', n)

agg = WindowAggregator(window_size=6)
for ev in events: agg.add(ev)
key = ('A', int(df['asset_id'].iloc[0]))
print('aggregate:', agg.aggregate(key))
```
